```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import accuracy_score
```

```python
df= pd.read_csv("/content/diabetes_012_health_indicators_BRFSS2015.csv")

df.head()
```

#Methods

```python
df.shape
df.info()
df.describe()
df.isnull()
```

#Target Value

```python


y=df['Diabetes_012']

x=df.drop('Diabetes_012', axis=1)
x.describe()
```

#Train and test select

```python
x_train, x_val, y_train, y_val = train_test_split(x, y, random_state=0)
```

#Logistic model


```python
Diabetes_012_model = LinearRegression()

Diabetes_012_model.fit(x_train, y_train)
```

#Predicted value

```python
Val_predictions = Diabetes_012_model.predict(x_val)
Val_mean=mean_absolute_error(y_val, Val_predictions)
print(f"On average, our model's predictions are off by:{Val_mean}")
```
